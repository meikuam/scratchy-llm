```python
import os
import requests
from pathlib import Path
```

```python
data_path = Path("../data")

warandpeace_path = data_path / "warandpeace.txt"
warandpeace_path.parent.mkdir(exist_ok=True, parents=True)

warandpeace_link = "https://gist.githubusercontent.com/romaklimenko/c95f3a864828f7f034b7a33d1676e420/raw/55f9027799b5b3c67e2f7cb3d6a7154f707ff08a/warandpeace.txt"
```

```python
response = requests.get(warandpeace_link)
with open(warandpeace_path, 'wb') as file:
    file.write(response.content)
```
